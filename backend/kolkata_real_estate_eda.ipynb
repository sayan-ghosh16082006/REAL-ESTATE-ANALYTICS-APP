## `Kolkata Real Estate Dataset — Univariate & Bivariate Analysis`

This notebook explores the `v2.csv` real-estate listings dataset (99acres-style scrape of Kolkata property listings).

**Contents**
1. Setup & Data Loading
2. Data Overview & Cleaning
3. Univariate Analysis (numerical & categorical)
4. Bivariate Analysis (numeric-numeric, numeric-categorical, categorical-categorical)
5. Key Takeaways


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 5)
pd.set_option('display.max_columns', 60)

df = pd.read_csv('D:/CODING/Virtual_py/PROJECTS/real_estate_ml/backend/kolkata.csv/v2.csv')
print(df.shape)
df.head()
```

### 2. Data Overview & Cleaning

```python
df.info()
```

```python
# Missing values
missing = df.isnull().sum()
missing = missing[missing > 0].sort_values(ascending=False)
missing_pct = (missing / len(df) * 100).round(2)
pd.DataFrame({'missing_count': missing, 'missing_pct': missing_pct})
```

```python
# Select key columns relevant for analysis
num_cols = ['PRICE_Cr', 'AREA_SQFT', 'PRICE_PER_SQFT(INR)', 'BEDROOM_NUM',
            'TOTAL_FLOOR', 'AGE', 'BALCONY_NUM', 'TOTAL_LANDMARK_COUNT']

cat_cols = ['PREFERENCE', 'OWNTYPE_label', 'FACING_LABEL', 'FURNISH_LABEL',
            'PROPERTY_type', 'CITY_LABELS']

df[num_cols].describe().T
```

```python
# Basic cleaning: drop rows with missing price/area for price-based analysis
df_clean = df.copy()
df_clean = df_clean[df_clean['PRICE_Cr'].notnull() & df_clean['AREA_SQFT'].notnull()]

# Remove obvious outliers using the 1st-99th percentile for core numeric columns
for c in ['PRICE_Cr', 'AREA_SQFT', 'PRICE_PER_SQFT(INR)']:
    low, high = df_clean[c].quantile([0.01, 0.99])
    df_clean = df_clean[(df_clean[c] >= low) & (df_clean[c] <= high)]

print(f"Original rows: {len(df)}  |  Rows after cleaning: {len(df_clean)}")
```

### 3. Univariate Analysis

#### 3.1 Numerical Variables

```python
fig, axes = plt.subplots(2, 3, figsize=(18, 10))
axes = axes.flatten()

plot_cols = ['PRICE_Cr', 'AREA_SQFT', 'PRICE_PER_SQFT(INR)', 'BEDROOM_NUM', 'TOTAL_FLOOR', 'AGE']

for ax, col in zip(axes, plot_cols):
    sns.histplot(df_clean[col].dropna(), kde=True, ax=ax, color='steelblue')
    ax.set_title(f'Distribution of {col}')

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(2, 3, figsize=(18, 8))
axes = axes.flatten()

for ax, col in zip(axes, plot_cols):
    sns.boxplot(x=df_clean[col].dropna(), ax=ax, color='lightcoral')
    ax.set_title(f'Boxplot of {col}')

plt.tight_layout()
plt.show()
```

```python
df_clean[plot_cols].describe().T
```

**Observations:**
- `PRICE_Cr` and `PRICE_PER_SQFT(INR)` are right-skewed — most listings cluster at the lower end with a long tail of premium properties.
- `AREA_SQFT` is also right-skewed, dominated by small-to-mid-sized apartments.
- `BEDROOM_NUM` is discrete and dominated by 2–3 BHK units.
- `AGE` (freshness label / age code) and `TOTAL_FLOOR` show multi-modal patterns tied to specific building types.


#### 3.2 Categorical Variables

```python
fig, axes = plt.subplots(2, 3, figsize=(20, 11))
axes = axes.flatten()

for ax, col in zip(axes, cat_cols):
    order = df[col].value_counts().index
    sns.countplot(data=df, y=col, order=order, ax=ax, palette='viridis', hue=col, legend=False)
    ax.set_title(f'Count of {col}')
    ax.set_xlabel('Count')

plt.tight_layout()
plt.show()
```

```python
for col in cat_cols:
    print(f"--- {col} ---")
    print(df[col].value_counts(dropna=False, normalize=True).mul(100).round(1).astype(str) + '%')
    print()
```

**Observations:**
- `Cooperative Society` is by far the dominant ownership type.
- `East` and `North-West` are the most common facing directions.
- `Fully furnished` listings dominate the furnishing category.
- `Residential Apartment` is overwhelmingly the most common property type.
- `Kolkata South` and `Kolkata East` account for the bulk of listings.


### 4. Bivariate Analysis

#### 4.1 Numeric vs Numeric

```python
corr = df_clean[num_cols].corr(numeric_only=True)
plt.figure(figsize=(9, 7))
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm', center=0)
plt.title('Correlation Heatmap — Numerical Features')
plt.show()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

sns.scatterplot(data=df_clean, x='AREA_SQFT', y='PRICE_Cr', alpha=0.4, ax=axes[0])
axes[0].set_title('Area vs Price')

sns.scatterplot(data=df_clean, x='AREA_SQFT', y='PRICE_PER_SQFT(INR)', alpha=0.4, ax=axes[1], color='darkorange')
axes[1].set_title('Area vs Price per Sqft')

sns.scatterplot(data=df_clean, x='BEDROOM_NUM', y='PRICE_Cr', alpha=0.4, ax=axes[2], color='seagreen')
axes[2].set_title('Bedrooms vs Price')

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

sns.boxplot(data=df_clean, x='FURNISH_LABEL', y='PRICE_Cr', ax=axes[0], hue='FURNISH_LABEL', legend=False, palette='pastel')
axes[0].set_title('Price by Furnishing Status')

sns.boxplot(data=df_clean, x='BEDROOM_NUM', y='AREA_SQFT', ax=axes[1], hue='BEDROOM_NUM', legend=False, palette='coolwarm')
axes[1].set_title('Area by Bedroom Count')
axes[1].set_xlim(-1, 8)

plt.tight_layout()
plt.show()
```

```python
df_clean.groupby('CITY_LABELS')[['PRICE_Cr', 'AREA_SQFT', 'PRICE_PER_SQFT(INR)']].median().sort_values('PRICE_Cr', ascending=False)
```

**Observations:**
- Median prices vary meaningfully across city zones, with some zones commanding a clear premium.
- Fully furnished listings tend to command a modest price premium over unfurnished ones.
- Area increases with bedroom count as expected, but with growing variance for larger configurations.


#### 4.3 Categorical vs Categorical

```python
ct = pd.crosstab(df['CITY_LABELS'], df['PROPERTY_type'])
ct_pct = pd.crosstab(df['CITY_LABELS'], df['PROPERTY_type'], normalize='index').mul(100).round(1)

plt.figure(figsize=(11, 6))
sns.heatmap(ct_pct, annot=True, fmt='.1f', cmap='Blues')
plt.title('Property Type Share (%) by City Zone')
plt.show()

ct
```

### 5. Key Takeaways

- The dataset is dominated by **Residential Apartments** in **Cooperative Society** ownership, mostly **2–3 BHK** configurations.
- **Price and Area are strongly positively correlated**, while price-per-sqft is more strongly driven by location and property type than by area alone.
- **Kolkata South and East** are the largest sub-markets by listing volume, and price levels differ meaningfully by city zone.
- Furnishing status and facing direction show some relationship with price, though city zone and area remain the dominant price drivers.
- Several columns (`BALCONY_NUM`, `AMENITIES`, `TRANSACT_TYPE`) have notable missingness and should be treated carefully in any downstream modeling.

```python
def IQR_treatment(column_name):
    upper = column_name.quantile(0.99)
    lower = column_name.quantile(0.01)


    return np.where(
        column_name > upper,
        upper,
        np.where(
            column_name < lower,
            lower,
            column_name
        )
    )
```

```python
import re
import pandas as pd

def extract_amenities(text):
    if pd.isna(text):
        return None
    # Regex to capture text after 'amenities like' until the period
    match = re.search(r'amenities like (.*?)(?:\.|$)', text, flags=re.IGNORECASE)
    if match:
        return match.group(1).strip()
    return np.nan

```
