```python
import pandas as pd
import re

def convert_price_to_cr(price_str):
    if pd.isna(price_str):
        return None
    
    price_str = str(price_str).replace(",", "").strip()
    
    # Handle ranges (take average)
    if "-" in price_str:
        parts = price_str.split("-")
        values = [convert_price_to_cr(p.strip()) for p in parts]
        return sum(values) / len(values) if all(values) else None
    
    # Crores
    if "Cr" in price_str:
        num = re.findall(r"[\d\.]+", price_str)
        return float(num[0]) if num else None
    
    # Lakhs
    if "L" in price_str:
        num = re.findall(r"[\d\.]+", price_str)
        return float(num[0]) / 100 if num else None
    
    # Per Bedroom / Bed (rental values → convert to Cr)
    if "/Bedroom" in price_str or "/Bed" in price_str:
        num = re.findall(r"[\d\.]+", price_str)
        return float(num[0]) / 1e7 if num else None
    
    # Plain number (₹)
    num = re.findall(r"[\d\.]+", price_str)
    return float(num[0]) / 1e7 if num else None

# Apply to your DataFrame
df["PRICE_Cr"] = df["PRICE"].apply(convert_price_to_cr)

print(df[["PRICE", "PRICE_Cr"]].head(20))
```

```python
df["AREA"].value_counts()

def convert_area(area_str):
    if pd.isna(area_str):
        return None
    
    # Remove commas and unit
    area_str = str(area_str).replace(",", "").replace("sq.ft.", "").strip()
    
    # Handle ranges
    if "-" in area_str:
        parts = re.findall(r"[\d\.]+", area_str)
        if len(parts) == 2:
            return (float(parts[0]) + float(parts[1])) / 2
        else:
            return None
    
    # Single value
    num = re.findall(r"[\d\.]+", area_str)
    return float(num[0]) if num else None

# Apply to your DataFrame
df["AREA_SQFT"] = df["AREA"].apply(convert_area)
```

```python
# Assuming PRICE_Cr is your cleaned price column (in Crores)
# Convert Crores back to INR for per sqft calculation
df["PRICE_INR"] = df["PRICE_Cr"] * 1e7   # 1 Cr = 10,000,000 INR

# Recalculate price per unit area (₹ per sqft)
df["PRICE_PER_SQFT"] = df["PRICE_INR"] / df["AREA_SQFT"]

# print(df[["PRICE", "PRICE_Cr", "AREA_SQFT", "PRICE_PER_UNIT_AREA"]].head(10))
df.sample(5)
```
