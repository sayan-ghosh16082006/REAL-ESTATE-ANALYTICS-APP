```python
import ast

val = df["FORMATTED_LANDMARK_DETAILS"][1]
parsed = ast.literal_eval(val) 
print(parsed)
```

```python
def extract_category(val):
    # First handle missing values
    if pd.isna(val):
        return ""
    
    # Then safely parse the string
    try:
        parsed = ast.literal_eval(val)
    except (ValueError, SyntaxError):
        return ""
    
    # Handle dict case
    if isinstance(parsed, dict):
        return parsed.get("category", "")
    
    # Handle list of dicts case
    if isinstance(parsed, list):
        return [item.get("category", "") for item in parsed if isinstance(item, dict)]
    
    return ""

# print(extract_category(df["FORMATTED_LANDMARK_DETAILS"][1]))

df["LANDMARK_DETAILS"] = df["FORMATTED_LANDMARK_DETAILS"].apply(extract_category)
```

```python
for i in range(df["location"].shape[0]):
    parsed = ast.literal_eval(df["location"][i])
    print(parsed)
```

```python
def extract_city_num(val):
    if pd.isna(val):
        return ""
    
    # Then safely parse the string
    try:
        parsed = ast.literal_eval(val)
    except (ValueError, SyntaxError):
        return ""

    for i in range(df["location"].shape[0]):
        return parsed.get("CITY")

def extract_locality_name(val):
    if pd.isna(val):
        return ""
    
    # Then safely parse the string
    try:
        parsed = ast.literal_eval(val)
    except (ValueError, SyntaxError):
        return ""

    for i in range(df["location"].shape[0]):
        return parsed.get("LOCALITY_NAME")


```

```python
print(extract_city_num(df["location"][7]))
print(extract_locality_name(df["location"][7]))
```

```python
df["CITY_NUM"] = df["location"].apply(extract_city_num)
df["LOCALITY"] = df["location"].apply(extract_locality_name)
```
