```python
import ast

def extract_list(val):
    if pd.isna(val):
        return []
    try:
        parsed = ast.literal_eval(val)   # safely evaluate the string
        if isinstance(parsed, list):
            return parsed                # already a list
        else:
            return [parsed]              # wrap single values
    except (ValueError, SyntaxError):
        return []
```

```python
description_df["LANDMARK_DETAILS"] = description_df["LANDMARK_DETAILS"].apply(extract_list)
```

```python
description_df["SECONDARY_TAGS"] = description_df["SECONDARY_TAGS"].apply(extract_list)

description_df["SECONDARY_TAGS_STR"] = description_df["SECONDARY_TAGS"].apply(" ".join)
```

```python
# Join relevant text columns into one string per row
description_df["FACILITIES_TEXT"] = (
    description_df["DESCRIPTION"].fillna("") + " " +
    description_df["PROP_HEADING"].fillna("") + " " +
    description_df["SECONDARY_TAGS_STR"].fillna("") + " " +
    description_df["LANDMARK_DETAILS_STR"].fillna("") + " " +
    description_df["FURNISH_LABEL"].fillna("") + " " +
    description_df["PROPERTY_type"].fillna("") + " " +
    description_df["OWNTYPE_label"].fillna("")
)
```

```python
description_df.iloc[1].FACILITIES_TEXT
```

```python
# Apply TF-IDF
vectorizer = TfidfVectorizer(max_features=5000, stop_words='english', ngram_range=(1,2))
tfidf_matrix = vectorizer.fit_transform(description_df["FACILITIES_TEXT"])
```

```python
tfidf_matrix.toarray().shape
```

```python
cosine_sim1 = cosine_similarity(tfidf_matrix, tfidf_matrix)
```

```python
cosine_sim1.shape
```

```python
def recommend_properties(property_name, cosine_sim = cosine_sim1):
    # get the index of the property
    idx = description_df.index[description_df["PROP_NAME"] == property_name].tolist()[0]

    # get the pairwise similarity scores for the property
    sim_scores = list(enumerate(cosine_sim[idx]))

    #sort the properties based on similarity scores
    sim_scores = sorted(sim_scores, key=lambda x : x[1], reverse = True)

    # get the scores of the top 10 properties
    sim_scores = sim_scores[1:11]

    # get the property indices
    property_indices = [i[0] for i in sim_scores]

    recommendation_df = pd.DataFrame({
        "Property Name" : description_df["PROP_NAME"].iloc[property_indices],
        "Similarity Scores" : sim_scores
    })

    return recommendation_df
```

```python
price_df = df[["PROP_NAME", "AGE", "BEDROOM_NUM", "BALCONY_NUM", "AREA_SQFT","PRICE_Cr"]]
```

```python
price_df.head()
```

```python
from sklearn.preprocessing import StandardScaler

# Numeric columns to scale
numeric_cols = ["AGE", "BEDROOM_NUM", "BALCONY_NUM", "AREA_SQFT", "PRICE_Cr"]

# Scale numeric features
scaler = StandardScaler()
scaled_values = scaler.fit_transform(price_df[numeric_cols])

# Build scaled DataFrame
scaled_df = pd.DataFrame(scaled_values, columns=numeric_cols, index=price_df.index)



```

```python
scaled_df.head()
```

```python
cosine_sim_price = cosine_similarity(scaled_df, scaled_df)
```

```python
cosine_sim_price.shape
```

```python
def recommend_properties_with_price(property_name):

    idx = price_df.index[price_df["PROP_NAME"] == property_name].tolist()[0]

    sim_scores = list(enumerate(cosine_sim_price[idx]))
    sim_scores = sorted(sim_scores, key = lambda x : x[1], reverse = True)
    sim_scores = sim_scores[1:11]

    property_indices = [i[0] for i in sim_scores]

    recomended_df = pd.DataFrame({
        "Property Name" : price_df["PROP_NAME"].iloc[property_indices],
        "Similarity Scores" : sim_scores
    })

    return recomended_df
```

```python
location_df = df[["PROP_NAME", "LATITUDE", "LONGITUDE","CITY_LABELS", "LOCALITY"]]
```

```python
location_df.head()
```

```python
scaler = StandardScaler()
coords_scaled = scaler.fit_transform(location_df[["LATITUDE", "LONGITUDE"]])

```

```python
# Compute similarity
cosine_sim_location = cosine_similarity(coords_scaled)

```

```python
cosine_sim_location.shape
```

```python
def hybrid_recommend_properties_by_location(property_name, w_price=0.4, w_landnark=0.3, w_location=0.3,top_n=10):

    idx = location_df.index[location_df["PROP_NAME"] == property_name].tolist()[0]

    price_scores = cosine_sim_price[idx]
    landnark_scores = cosine_sim1[idx]
    location_scores = cosine_sim_location[idx]

    final_scores = (w_price * price_scores +
                    w_landnark * landnark_scores +
                    w_location * location_scores)
    
    sim_scores = list(enumerate(final_scores))
    sim_scores = sorted(sim_scores, key=lambda x: x[1], reverse=True)[1:top_n+1]

    property_indices = [i[0] for i in sim_scores]
    scores = [i[1] for i in sim_scores]

    recommended_df = pd.DataFrame({
    "Property Name": description_df["PROP_NAME"].iloc[property_indices].values,
    "Hybrid Score": scores
    }, index=property_indices)


    return recommended_df
```

```python
rec_df = hybrid_recommend_properties_by_location("team taurus kabya", 0, 0, 1)

# Extract indices from the Similarity Scores column
indices = rec_df.index.tolist()

# Use those indices to get full records
actual_records = df.iloc[indices][["PROP_NAME", "DESCRIPTION", "AGE", "PROP_HEADING", "SECONDARY_TAGS", "LATITUDE","LONGITUDE","OWNTYPE_label","PRICE_Cr", "AREA_SQFT", "FURNISH_LABEL", "PROPERTY_type","CITY_LABELS", "LANDMARK_DETAILS", "LOCALITY"]]

actual_records.head()
```
